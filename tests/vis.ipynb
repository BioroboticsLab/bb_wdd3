```python
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
```

```python
img1 = np.array(Image.open("aug_test_aug_none/frame_007.png"))
img2 = np.array(Image.open("aug_test_aug_rotate/frame_007.png"))

fig, axes = plt.subplots(1, 2, figsize=(13, 7))

axes[0].imshow(img1)
axes[0].axis('off')

axes[1].imshow(img2)
axes[1].axis('off')


plt.tight_layout()
plt.show()
```
